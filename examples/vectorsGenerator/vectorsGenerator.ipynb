```python
import os
path = "/home/stefanovjc/Documents"
os.chdir(path)
%matplotlib qt
from geompy import *
import random
```

```python
for jj in range(100):
    
    xintervals = [(3, 5), (-5, -3)]
    yintervals = [(2, 4), (-4, -2)]
    
    # --- PUNTO A (x e y indipendenti) ---
    x_a_scelto = random.choice(xintervals)
    ax = random.uniform(x_a_scelto[0], x_a_scelto[1])
    
    y_a_scelto = random.choice(yintervals)
    ay = random.uniform(y_a_scelto[0], y_a_scelto[1])
    
    
    # --- PUNTO B (x e y indipendenti da A e tra di loro) ---
    x_b_scelto = random.choice(xintervals)
    bx = random.uniform(x_b_scelto[0], x_b_scelto[1])
    
    y_b_scelto = random.choice(yintervals)
    by = random.uniform(y_b_scelto[0], y_b_scelto[1])
    
    
    v1, v2 = vector(), vector()
    O = point(0, 0)
    C = point(3, 0, draw = False)
    O.step = 1
    v1.points = v2.points = O
    
    A, B = point(ax, ay), point(bx, by)
    v1.points, v2.points  = A, B
    
    alpha, beta = angle(), angle()
    
    alpha.centre = beta.centre = O
    alpha.points = beta.points = C
    alpha.points, beta.points = v1.points[1], v2.points[1]
    alpha.radius = 1.2
    
    alpha.color = v1.color
    beta.color = v2.color
    
    alpha.name = rf"$\alpha = {alpha.size*180/pi:.2f}^\circ$"
    beta.name = rf"$\beta = {beta.size*180/pi:.2f}^\circ$"
    
    v1text = text()
    v2text = text()
    j = 1
    
    x = v1.points[j].x/2
    y = v1.points[j].y*1.4/2
    v1text.x, v1text.y = x, y
    
    x = v2.points[j].x/2
    y = v2.points[j].y*1.4/2
    v2text.x, v2text.y = x, y
    
    v1text.color, v2text.color = v1.color, v2.color
    
    v1text.name = r"$\vec{v}_1$"
    v2text.name = r"$\vec{v}_2$"
    
    v1textI, v2textI = text(), text()
    v1textI.x, v1textI.y = 6, 5
    v2textI.x, v2textI.y = 6, 4
    v1textI.name = rf"$v_1 = {(v1.x[1]**2 + v1.y[1]**2)**.5:.2f}$"
    v2textI.name = rf"$v_2 = {(v2.x[1]**2 + v2.y[1]**2)**.5:.2f}$"
    v1textI.color = v1text.color
    v2textI.color = v2text.color
    pathImage = path + "/pictures/"
    stringa = f"{jj:02d}"
    
    plt.savefig(pathImage + stringa + "_graph.png")
```
